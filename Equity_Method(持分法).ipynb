```python
# =========================================
# Basic Equity Method Model (1 Parent Company + 1 Associate)
# =========================================
import pandas as pd

# -----------------------------------------
# 1. Initial Data Setup
# -----------------------------------------
# Assume the parent company owns 40% of the associate's shares
ownership = 0.4  # Ownership ratio 40%

# Parent company's investment account (investment in associate)
parent_investment = 1000  # Investment amount (at acquisition)

# Associate's net income for the period (e.g., 1 year)
associate_net_income = 500  # Net income

# Dividends paid by the associate to the parent company
dividend_paid = 100  # Dividends (cash payment)

# -----------------------------------------
# 2. Recognize Parent's Share of Profit
# -----------------------------------------
# Associate's net income × ownership ratio = Parent's share of profit
equity_income = associate_net_income * ownership

# Parent increases the investment account by this profit (investment value rises)
parent_investment += equity_income

print(f"Parent company's investment account (after recognizing equity income): {parent_investment}")

# -----------------------------------------
# 3. Handling Dividends Received
# -----------------------------------------
# Dividends are "cash received from the investee,"
# so the investment account is reduced by that amount (return of investment)
parent_investment -= dividend_paid * ownership

print(f"Investment account after dividend adjustment: {parent_investment}")

# -----------------------------------------
# 4. Parent's Balance Sheet Display After Consolidation
# -----------------------------------------
# (In practice, we adjust the investment account rather than including associate's assets and liabilities)
parent_bs = {
    "Cash": 500 + dividend_paid * ownership,  # Cash increases by dividend received
    "Investment in Associate": parent_investment,
    "Total Equity": 0  # Will calculate later
}

# Total Equity = Total Assets
parent_bs["Total Equity"] = parent_bs["Cash"] + parent_bs["Investment in Associate"]

# Convert to DataFrame for display
df_bs = pd.DataFrame(parent_bs, index=["Parent Company (After Equity Method)"])
print("\nParent Company's Balance Sheet (After Equity Method):")
print(df_bs)

# -----------------------------------------
# 5. Summary (Explanation)
# -----------------------------------------
print("\n【Process Summary】")
print("① Add 40% (200) of associate's net income 500 to the parent's investment account.")
print("② Receive 40% (40) of dividends 100 from the associate and reduce investment account by 40.")
print("③ Result: Investment account increases by +160 (1000 → 1160).")
```

```python
# =========================================
# 持分法の基本モデル（親会社1社＋関連会社1社）
# =========================================
import pandas as pd

# -----------------------------------------
# 1. 初期データ設定
# -----------------------------------------
# 親会社が関連会社株式を40%保有していると仮定します
ownership = 0.4  # 持分比率 40%

# 親会社の投資勘定（関連会社株式）
parent_investment = 1000  # 投資額（取得時点）

# 関連会社の当期純利益（例：1年間）
associate_net_income = 500  # 純利益

# 関連会社が親会社に配当を支払った金額
dividend_paid = 100  # 配当金（現金支払）

# -----------------------------------------
# 2. 親会社の持分に応じた利益の認識
# -----------------------------------------
# 関連会社の純利益 × 持分比率 = 親会社の持分利益
equity_income = associate_net_income * ownership

# 親会社はこの利益分だけ投資勘定を増やす（投資の価値が上がるため）
parent_investment += equity_income

print(f"親会社の投資勘定（持分利益反映後）: {parent_investment}")

# -----------------------------------------
# 3. 配当金を受け取った場合の処理
# -----------------------------------------
# 配当金は「投資先から現金を受け取る」ので、
# 投資勘定をその分だけ減額します（投資の回収）
parent_investment -= dividend_paid * ownership

print(f"配当金処理後の投資勘定: {parent_investment}")

# -----------------------------------------
# 4. 連結時の親会社BS上の表示イメージ
# -----------------------------------------
# （実務では、子会社の資産負債を取り込むのではなく、投資勘定を調整）
parent_bs = {
    "現金": 500 + dividend_paid * ownership,  # 配当金受取分、現金が増える
    "関連会社株式（投資勘定）": parent_investment,
    "純資産合計": 0  # 後で計算
}

# 純資産 = 資産合計
parent_bs["純資産合計"] = parent_bs["現金"] + parent_bs["関連会社株式（投資勘定）"]

# 表示用にDataFrameに変換
df_bs = pd.DataFrame(parent_bs, index=["親会社（持分法適用後）"])
print("\n親会社の貸借対照表（持分法適用後）:")
print(df_bs)

# -----------------------------------------
# 5. サマリー（説明出力）
# -----------------------------------------
print("\n【処理の流れ】")
print("① 関連会社の純利益500のうち40%（200）を親会社の投資勘定に加算。")
print("② 関連会社からの配当金100のうち40%（40）を受け取り、投資勘定を40減額。")
print("③ 結果として投資勘定は +160増加（1000→1160）。")
```
